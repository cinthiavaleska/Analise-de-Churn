```python
#IMPORTACAO DAS BIBLIOTECAS

import pandas as pd
import numpy as np
```

```python
#IMPORTACAO CVS
df_customers = pd.read_csv("customers_raw.csv")
df_customers.head()
```

```python
#DIAGNOSTICO DE QUALIDADE DOS DADOS
df_customers.info()
```

```python
#DIAGNOSTICAR QUALIDADE DOS DADOS
#VERIFICAR VALORES NULOS
df_customers.isnull().sum()
```

```python
# TRATAMENTO DOS VALORES NULOS

df_customers["city"] = df_customers["city"].fillna("Não informado")
df_customers["service_type"] = df_customers["service_type"].fillna("Não informado")
df_customers["contract_type"] = df_customers["contract_type"].fillna("Não informado")
df_customers["payment_method"] = df_customers["payment_method"].fillna("Não informado")

# VALIDAR
df_customers.isnull().sum()
```

```python
#PADRONIZAR COLUNAS 

colunas_texto = [
    "company_name",
    "industry",
    "company_size",
    "state",
    "city",
    "acquisition_channel",
    "service_type",
    "contract_type",
    "payment_method"
]

for coluna in colunas_texto:
    df_customers[coluna] = df_customers[coluna].str.strip().str.title()

df_customers.head(10)
```

```python
#CORRIGIR TIPO DE COLUNA
df_customers["customer_since"] = pd.to_datetime(
    df_customers["customer_since"],
    format="mixed",
    dayfirst=True
)
#VALIDAR
df_customers.info()
```
