```python
#IMPORTACAO DAS BIBLIOTECAS

import pandas as pd
```

```python
#IMPORTACAO CVS
df_monthly = pd.read_csv("../data/raw/customer_monthly_raw.csv")
df_monthly.head()
```

```python
#EXPLORACAO DOS DADOS
df_monthly.info()
```

```python
#VERIFICACAO DE VALORES NULOS
df_monthly.isnull().sum()
```

```python
# TRATAMENTO DE NULOS
df_monthly["meetings_count"] = df_monthly["meetings_count"].fillna(0)

df_monthly["nps"] = df_monthly["nps"].fillna(df_monthly["nps"].median())

df_monthly["satisfaction_score"] = df_monthly["satisfaction_score"].fillna(df_monthly["satisfaction_score"].median())

df_monthly["service_quality_score"] = df_monthly["service_quality_score"].fillna(df_monthly["service_quality_score"].median())

#VALIDAR NULOS
df_monthly.isnull().sum()
```

```python
#VERIFICAR DUPLICADOS
df_monthly.duplicated().sum()
```

```python
#REMOVER DUPLICADOS

df_monthly = df_monthly.drop_duplicates()

# VERIFICANDO DUPLICADOS POR CLIENTE E MÊS
df_monthly.duplicated(
    subset=["customer_id", "reference_month"]
).sum()
```

```python
#CORRIGIR TIPO DE COLUNA
df_monthly["reference_month"] = pd.to_datetime(
    df_monthly["reference_month"],
    format="mixed",
    dayfirst=True
)
#VALIDAR TIPOS
df_monthly.info()
```

```python
#PADRONIZAR COLUNAS E REMOVER ESPACOS


df_monthly["customer_id"] = df_monthly["customer_id"].str.strip()

df_monthly["customer_status"] = (
    df_monthly["customer_status"]
    .str.strip()
    .str.title()
)


# VALIDAR COLUNAS

df_monthly.head()
```

```python
#VALIDACAO FINAL
df_monthly.info()
```
